```python
from typing import TypedDict
from langgraph.graph import StateGraph,START , END
```

```python
class SpeedState(TypedDict):
    distance:float
    time:float
    speed:float
```

```python
def calculate_speed(state:SpeedState):
    distance = state['distance']
    time = state['time']
    speed = distance/time
    state['speed'] = speed
    return state
```

```python
# creation of graph
graph = StateGraph(SpeedState)

# add nodes in the graphs
graph.add_node('calculate_speed',calculate_speed)

# add edges in the graphs
graph.add_edge(START,"calculate_speed")

graph.add_edge("calculate_speed",END)

# lets compile it 
workflow = graph.compile()

# lets execute it
initial_state : SpeedState = {
    "distance":100.0,
    "time":5.0,
    "speed":0.0
    }

final_state = workflow.invoke(initial_state)

print(final_state)
```
