```python
from typing import Annotated
from typing_extensions import TypedDict
import operator
from langgraph.graph import StateGraph, START, END

class WorkflowState(TypedDict):
    text: str
    outputs: Annotated[list[str], operator.add]
    final: str

def sentiment_branch(state: WorkflowState):
    text = state["text"]
    return {"outputs": [f"Sentiment analyzed: {text}"]}


def keywords_branch(state: WorkflowState):
    text = state["text"]
    return {"outputs": [f"Keywords extracted: {text}"]}


def summarize(state: WorkflowState):
    return {
        "final": " | ".join(state["outputs"])
    }


builder = StateGraph(WorkflowState)

builder.add_node("sentiment", sentiment_branch)
builder.add_node("keywords", keywords_branch)
builder.add_node("summarize", summarize)

builder.add_edge(START, "sentiment")
builder.add_edge(START, "keywords")

builder.add_edge("sentiment", "summarize")
builder.add_edge("keywords", "summarize")

builder.add_edge("summarize", END)

graph = builder.compile()

result = graph.invoke(
    {
        "text": "LangGraph can run branches in parallel.",
        "outputs": []
    }
)

print(result["final"])
```
