```python
from langgraph.graph import StateGraph, START , END
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from typing import TypedDict

load_dotenv()
model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.9,
    max_tokens=4048,
)
```

# Research & Report Generator
            Topic
              ↓
            Plan
              ↓
 ┌──────────--┬──────────-┬──────────┐
 ↓            ↓           ↓
Research  Examples   Statistics
 └──────────--┴──────────-┴──────────┘
              ↓
           Analysis
              ↓
          Write Report



```python
class ResRepGen(TypedDict):
    topic : str
    plan : str 
    research: str
    examples: str
    stats: str
    analysis: str
    report: str
```

```python
def plan(state:ResRepGen)->ResRepGen:
    # 1. Get the topic from the current state
    topic = state['topic']
    # 2. Create a prompt using the topic
    prompt = f"Create a plan for this {topic}"
    # 3. Send the prompt to the model and get the generated plan
    plan = str(model.invoke(prompt))
    # 4. Store the generated plan in the state
    state['plan'] = plan
    # 5. Return the updated state
    return state
```

```python
def research(state:ResRepGen)->ResRepGen:
    plan = state['plan']
    prompt = f"write a research based on this {plan}"
    research = str(model.invoke(prompt))
    state["research"] = research
    return state
```

```python
def examples(state:ResRepGen)->ResRepGen:
    plan = state['plan']
    prompt = f"Give me examples based on the {plan}"
    examples = str(model.invoke(prompt))
    state['examples'] = examples
    return state
```

```python
def stats(state:ResRepGen)->ResRepGen:
    plan = state['plan']
    prompt = f"Give me statistics about {plan}"
    stats = str(model.invoke(prompt))
    state["stats"] = stats
    return state
```

```python
def analysis(state:ResRepGen)->ResRepGen:
    research = state['research']
    examples = state['examples']
    stats = state['stats']

    prompt = f" Do an analysis by using {research} , {examples} and {stats}"
    analysis = str(model.invoke(prompt))
    state['analysis'] = analysis
    return state
```

```python
def report(state:ResRepGen)->ResRepGen:
    analysis = state['analysis']
    prompt = f"Create a report based on the {analysis}"
    report = str(model.invoke(prompt))
    state['report'] = report
    return state
```

```python
naveed = StateGraph(ResRepGen)

naveed.add_node("plan",plan)
naveed.add_node("research",research)
naveed.add_node("examples",examples)
naveed.add_node("stats",stats)
naveed.add_node("analysis",analysis)
naveed.add_node("report",report)
```

```python
naveed.add_edge(START,"plan")

naveed.add_edge("plan","research")
naveed.add_edge("plan","examples")
naveed.add_edge("plan","stats")

naveed.add_edge("research","analysis")
naveed.add_edge("examples","analysis")
naveed.add_edge("stats","analysis")

naveed.add_edge("analysis","report")
naveed.add_edge("report",END)
```

```python
workflow = naveed.compile()
```

```python
initial_state : ResRepGen = {
    "topic":"Can We learn and build AI Agents in 2 months?",
    "plan":"",
    "research":"",
    "examples":"",
    "stats":"",
    "analysis":"",
    "report":""
}

final_state = workflow.invoke(initial_state)
print(final_state["report"])
```
