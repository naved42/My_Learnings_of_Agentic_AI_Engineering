```python
from langgraph.graph import StateGraph, START , END
from typing import Literal, TypedDict

class LoanState(TypedDict):
    salary:int
    decision: Literal["approved", "rejected"]
    message: str




def route_loan(state:LoanState)->Literal['approve','reject']:
    salary = state['salary']

    if salary > 75000:
        return "approve"
    return "reject"


def loan_approved(state:LoanState):
    return {"decision":"approved",
            "message": f"Congrats! Loan approved for {state['salary']}"
}

def loan_rejected(state:LoanState):
    return {"decision":"rejected",
            "message": f"Alas! Loan rejected for {state['salary']}"
}



```

```python
builder = StateGraph(LoanState)

builder.add_node("approve",loan_approved)
builder.add_node("reject",loan_rejected)


builder.add_conditional_edges(START,route_loan)
builder.add_edge("approve",END)
builder.add_edge("reject",END)


workflow = builder.compile()

initial_state : LoanState ={
    "salary":505000
}
final_state = workflow.invoke(initial_state)
print(final_state)
```
